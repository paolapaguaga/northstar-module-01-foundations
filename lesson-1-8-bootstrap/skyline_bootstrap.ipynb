```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
skyline = pd.read_csv("../lesson-1-2-types-of-data/skyline_enrollments.csv")
data = skyline["hours_studied"].values
n = len(data)
B = 10000

np.random.seed(42)

boot_means = np.array([
    np.random.choice(data, size=n, replace=True).mean()
    for _ in range(B)
])

boot_medians = np.array([
    np.median(np.random.choice(data, size=n, replace=True))
    for _ in range(B)
])

boot_95th_percentile = np.array([
    np.percentile(np.random.choice(data, size=n, replace=True), 95)
    for _ in range(B)
])

ci_mean = np.percentile(boot_means, [2.5, 97.5])
ci_median = np.percentile(boot_medians, [2.5, 97.5])
ci_95th = np.percentile(boot_95th_percentile, [2.5, 97.5])

print(f"Sample size: {n} ")
print(f"Sample mean: {data.mean():.0f}")
print(f"Sample median: {np.median(data):.0f}")
print(f"Sample 95th percentile: {np.percentile(data, 95):.0f}")
print()
print(f"Bootstrap 95% CI for the mean:   ({ci_mean[0]:.2f}, {ci_mean[1]:.2f})")
print(f"Bootstrap 95% CI for the median: ({ci_median[0]:.2f}, {ci_median[1]:.2f})")
print(f"Bootstrap 95% CI for the 95th percentile: ({ci_95th[0]:.2f}, {ci_95th[1]:.2f})")
```

Part A: The two CIs are similar and it's expected because the sample is symmetric 

Part B: The median's CI tells that the true population median is estimated fairly precisely, with only 2 point range of plausible values at 95% confidence.

Part C: The 95th percentile's CI tends to be wider because it's estimated from only the small handful of observations near the tail of the distribution. The median for the contraru gets the weight of data clustered near the center. 


```python
bootstrap_df = pd.DataFrame({
    "boot_mean": boot_means,
    "boot_median": boot_medians,
    "boot_95th_percentile": boot_95th_percentile,
})

fig_mean = px.histogram(
    bootstrap_df,
    x="boot_mean",
    nbins=50,
    title="Bootstrap Distribution: Mean Total Hours Studied",
)
fig_mean.add_vline(x=ci_mean[0], line_dash="dash", annotation_text="2.5%")
fig_mean.add_vline(x=ci_mean[1], line_dash="dash", annotation_text="97.5%")
fig_mean.show()

fig_median = px.histogram(
    bootstrap_df,
    x="boot_median",
    nbins=50,
    title="Bootstrap Distribution: Median Total Hours Studied",
)
fig_median.add_vline(x=ci_median[0], line_dash="dash", annotation_text="2.5%")
fig_median.add_vline(x=ci_median[1], line_dash="dash", annotation_text="97.5%")
fig_median.show()

fig_95th = px.histogram(
    bootstrap_df,
    x="boot_95th_percentile",
    nbins=50,
    title="Bootstrap Distribution: 95th Percentile Total Hours Studied",
)
fig_95th.add_vline(x=ci_95th[0], line_dash="dash", annotation_text="2.5%")
fig_95th.add_vline(x=ci_95th[1], line_dash="dash", annotation_text="97.5%")
fig_95th.show()
```
